## Exercise 0: Environment and libraries

```python
import sys
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelBinarizer
from sklearn.datasets import load_iris

print(f"Python version: {sys.version}")
print(f"TensorFlow version: {tf.__version__}")
```

## Exercise 1: Regression - Optimize

```python
# Neural network designed for regression
model = keras.Sequential()
model.add(layers.Input(shape=(5,)))
model.add(Dense(8, activation='sigmoid'))
model.add(Dense(4, activation='sigmoid'))
model.add(Dense(1, activation='linear'))

# TODO: Set up the optimization
model.compile(
    optimizer='adam',
    loss='mean_squared_error', # Common loss for regression
    metrics=['mean_absolute_error'] # Common metric for regression
)
print("Model compiled with adam optimizer, mean_squared_error loss, and mean_absolute_error metric")
```

## Exercise 2: Regression example

```python
# Load the dataset
url = "https://01.tomorrow-school.ai/api/content/root/public/subjects/ai/keras-2/auto-mpg.csv"
df = pd.read_csv(url)

# Preprocess the data
# Drop the columns: model year, origin, car name
df = df.drop(columns=['model year', 'origin', 'car name'])

# Replace '?' with NaN and drop rows
df = df.replace('?', np.nan)
df = df.dropna()

# Convert all columns to numeric (just in case)
df = df.apply(pd.to_numeric)

# Split train test without shuffling. Keep 20% for the test set.
X = df.drop(columns=['mpg'])
y = df['mpg']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Scale the data using Standard Scaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"X_train_scaled shape: {X_train_scaled.shape}")
print(f"First 5 rows of X_train_scaled:\n{X_train_scaled[:5]}")
print(f"First 5 rows of y_train:\n{y_train[:5]}")
print(f"X_test_scaled shape: {X_test_scaled.shape}")
print(f"First 5 rows of X_test_scaled:\n{X_test_scaled[:5]}")
print(f"First 5 rows of y_test:\n{y_test[:5]}")

# Train a neural network
model = Sequential()
model.add(layers.Input(shape=(X_train_scaled.shape[1],)))
model.add(Dense(30, activation='sigmoid'))
model.add(Dense(30, activation='sigmoid'))
model.add(Dense(1, activation='linear'))

model.compile(loss='mean_squared_error', optimizer='adam', metrics=['mean_absolute_error'])

# Train with increased epochs
history = model.fit(X_train_scaled, y_train, epochs=1000, batch_size=20, verbose=0)

# Evaluate
loss, mae = model.evaluate(X_test_scaled, y_test, verbose=0)
print(f"Mean Absolute Error on Test Set: {mae}")
```

## Exercise 3: Multi-classification - Softmax

```python
# Create the architecture for a multi-class neural network
model = keras.Sequential()

# Hidden layer 1: 16 neurons and sigmoid
model.add(layers.Input(shape=(5,)))
model.add(Dense(16, activation='sigmoid'))

# Hidden layer 2: 8 neurons and sigmoid
model.add(Dense(8, activation='sigmoid'))

# Output layer: Adapted to multi-classification (5 classes in example)
# Softmax activation for probabilities
model.add(Dense(5, activation='softmax'))

model.summary()
```

## Exercise 4: Multi-classification - Optimize

```python
# Optimize the neural network
model.compile(
    loss='categorical_crossentropy', # Loss for multi-class classification
    optimizer='adam',
    metrics=['accuracy']
)
print("Model compiled with categorical_crossentropy loss, adam optimizer, and accuracy metric")
```

## Exercise 5: Multi-classification example

```python
# Load the dataset from sklearn
iris = load_iris()
X = iris.data
y = iris.target

# Split train test. Keep 20% for the test set. Use random_state=1.
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# Scale the data using Standard Scaler
scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)

# Use LabelBinarizer for one hot encoding of the target
lb = LabelBinarizer()
y_train_multi_class = lb.fit_transform(y_train)
y_test_multi_class = lb.transform(y_test)

print(f"First 10 values of y_train_multi_class:\n{y_train_multi_class[:10]}")

# Train a neural network
model = Sequential()
model.add(layers.Input(shape=(X_train_sc.shape[1],)))
model.add(Dense(10, activation='sigmoid'))
model.add(Dense(3, activation='softmax')) # 3 classes for Iris

# Compile model
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

# Train
model.fit(X_train_sc, y_train_multi_class, epochs=1000, batch_size=20, verbose=0)

# Evaluate
loss, accuracy = model.evaluate(X_test_sc, y_test_multi_class, verbose=0)
print(f"Accuracy on Test Set: {accuracy}")
```
